```python
def get_customer(customer_id: str):
    customer_data = pd.read_csv(DATA_DIR+"/customers.csv")
    data = customer_data[customer_data['customer_id'] == customer_id]
    if data.empty:
        return None
    else:
        result = data.to_dict(orient='records')[0]
    return result
```

```python
from pathlib import Path

from langchain_community.document_loaders import (
    DirectoryLoader,
    TextLoader,
)
from langchain_core.documents import Document


DOCUMENTS_DIR = Path("../data/acme_logistics/documents")
```

```python
from pathlib import Path

def load_documents() -> list[Document]:
    """
    Charge les documents métier de la base de connaissances.

    Returns:
        Liste des documents LangChain.
    """

    loader = DirectoryLoader(
        str(DOCUMENTS_DIR),
        glob="*.md",
        loader_cls=TextLoader,
        loader_kwargs={
            "encoding": "utf-8",
        },
    )

    return loader.load()
```

```python
load_documents()
```
